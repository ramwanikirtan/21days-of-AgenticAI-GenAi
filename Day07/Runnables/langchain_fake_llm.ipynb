```python
from abc import ABC, abstractmethod
```

```python
class Runnable(ABC):
    @abstractmethod
    def invoke(self, input_data):
        pass
```

```python
import random


class FakeLLM(Runnable):
    def __init(self):
        print("FakeLLM created")

    def predict(self, prompt):
        response_list = [
    "Hello! How can I help you today?",
    "That’s a great question. Let me think…",
    "I don’t have the exact answer, but here’s what I found.",
    "Could you please clarify what you mean?",
    "Yes, that’s correct!",
    "No, I don’t think that’s the case.",
    "Interesting point — let me expand on that.",
    "I’m not sure, but I can suggest where to look.",
    "Sure! Here’s a step-by-step explanation.",
    "I understand. Let’s try another approach."
    ]
        
        return {"response": random.choice(response_list)}
    
    def invoke(self, input_data):
        return self.predict(input_data)
```

```python
llm = FakeLLM()
```

```python
llm = FakeLLM()
```

```python
llm.predict("What is the capital of France?")
```

```python
class FakePromptTemplate:
    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables = input_variables


    def format(self, input_dict):
        return self.template.format(**input_dict)
```

```python
template = FakePromptTemplate(
    template="Explain the significance of {topic} in {no_of_words} words.",
    input_variables=["topic", "no_of_words"]
)
```

```python
prompt = template.format({"topic": "Artificial Intelligence", "no_of_words": 50})
```

```python
model = FakeLLM()
```

```python
model.predict(prompt)
```

```python
class FakeLLMChain:
    def __init__(self,llm,prompt):
        self.llm = llm
        self. prompt = prompt
        
    def run(self,input_dict):
        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)
        return result["response"]

        
        
```

```python
template = FakePromptTemplate(
    template="Explain the significance of {topic} in {no_of_words} words.",
    input_variables=["topic", "no_of_words"]
)
```

```python
llm = FakeLLM()
```

```python
chain = FakeLLMChain(llm=llm, prompt=template)
```

```python
chain.run({"topic": "Artificial Intelligence", "no_of_words": 50})
```
