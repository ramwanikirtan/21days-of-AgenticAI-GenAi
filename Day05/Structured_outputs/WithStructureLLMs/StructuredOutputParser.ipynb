### Benefits of using a structured output parser
- Enforces a clear schema so outputs are predictable and machine-readable.  


```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from langchain.output_parsers import StructuredOutputParser,ResponseSchema
load_dotenv()
```

```python
model =  ChatGoogleGenerativeAI(model="gemini-1.5-flash")
```

```python
schema = [
    ResponseSchema(name="fact 1",description="a interesting fact 1 about the topic in one short statement"),
    ResponseSchema(name="fact 2",description="a interesting fact 2 about the topic in one short statement"),
    ResponseSchema(name="fact 3",description="a interesting fact 3 about the topic in one short statement")
]


parser = StructuredOutputParser.from_response_schemas(schema)

template1 = PromptTemplate(
    template="give me 3 interesting DARK SCARY facts about {topic}. /n {format_instructions}",
    input_variables=["topic"],
    partial_variables={"format_instructions":parser.get_format_instructions()}
)
chain = template1 | model |parser
result = chain.invoke({"topic":"Artificial Intelligence"})

print(result)
```
