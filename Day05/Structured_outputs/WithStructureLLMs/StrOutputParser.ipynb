StrOutputParser (LangChain)
- StrOutputParser is a simple output parser in LangChain that returns the raw model output unchanged as a Python string.
- Use case: when you need an OutputParser-compatible object but want to accept the model response verbatim (no parsing or transformation).
- Behavior: accepts a raw text input and outputs that same text; it does not validate, extract structured data, or modify the content.


```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
load_dotenv()
```

```python
model =  ChatGoogleGenerativeAI(model="gemini-1.5-flash") 
```

```python
### 1st Prompt = detailed report
template1 = PromptTemplate(
    template="write a detailed report about {topic}" ,
    input_variables=["topic"]
)

```

```python
## 2nd Prompt = summary report

template2 = PromptTemplate(
    template="Write a 5 line summary of the following text:/n {detailed_report}",
    input_variables=["detailed_report"]
)
```

```python
prompt1 = template1.invoke({"topic":"Artificial Intelligence"})
detailed_report = model.invoke(prompt1)

prompt2 = template2.invoke({"detailed_report":detailed_report})
summary_report = model.invoke(prompt2)
print("summary_report:",summary_report)
```
